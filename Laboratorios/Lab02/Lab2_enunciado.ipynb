### ¿Ya, pero esto tiene aplicaciones más allá de guardar información en casilleros?

Para el caso de imágenes podemos encontrar múltiples aplicaciones con la manipulación de los tensores y operando matemáticamente con ellos. Algunas de las aplicaciones más conocidas (y que aplicaremos) son las siguientes:

- **Obtener el negativo de una imagen**: Consiste en obtener el color complementario de una imagen, para esto debemos restar la imagen con la máxima intensidad que podemos encontrar en una imagen (o sea 255).
    
    
- **Pasar a escala de grises una imagen**: Los valores RGB se convierten a escala de grises mediante la fórmula NTSC:

    $$ imagen\_gris = 0.299 * Rojo + 0.587 * Verde + 0.114 * Azul $$
    
    Esta fórmula representa la percepción relativa de la persona promedio del brillo de la luz roja, verde y azul.
    
    
- **Mejora de contraste**: Son múltiples las técnicas que nos permiten mejorar el contraste de una imagen, pero, una técnica simple para modificar los contrastes consta en obtener un factor de corrección llamado F en base al contraste deseado (C). Luego, es aplicado en la diferencia entre la imagen y 128. De esta forma obtenemos R, que es la imagen con la mejora de contraste deseada.

    $$ F=259*(C+255)/(255*(259-C)) $$
    $$ R=F*(img-128)+128 $$
  
  
- **Convolución**: Consiste en recorrer una imagen por cada uno de sus canales utilizando una matriz que lleva por nombre Kernel. El kernel, examinará los conjuntos de pixeles que recorre, aplicando una multiplicación de los valores circundantes ,y sumando todos los valores generados de este producto para generar un nuevo pixel en el tensor de salida.

Como estamos trabajando con imágenes los outputs deben ser enteros, por esto se le aconseja utilizar `.astype(int)` para transformar los arrays de salida a un formato legible por matplotlib.
- [X] Es importante que para las operaciones `__add__` y `__sub__` implementen una saturación de las imágenes. Es decir, la suma o resta deben dar como valor máximo 255 y/o como valor mínimos mayores o iguales a 0.

> **Notas:**
- Pueden reutilizar el código implementado en las celdas anteriores para implementar los métodos `show` e `info` . Sin embargo, No invoquen directamente esas funciones.
- La idea es que la imagen contenida en la clase sea inmutable, por ende, todos los metodos que modifiquen la imagen contenida en el objeto deberan retornar un nuevo objeto de la clase `Imagen` que contenga la imagen modificada.

````python
class Imagen:
    """Clase contenedora de imágenes"""

    def __init__(self, img):
        if isinstance(img, np.ndarray):
            if img.ndim != 3:
                raise ("El argumento debe ser un arreglo de numpy de solo 3 dimensiones")
            if img.shape[-1] != 3:
                raise (
                    "El argumento debe ser un arreglo de numpy de solo 3 dimensiones "
                    "tal que la última dimensión solo tiene 3 canales"
                )
            self.imagen = img
        else:
            raise TypeError(
                "Debes entregar un arreglo de numpy como argumento del constructor de "
                "Imagen"
            )

    def show(self):
        """Muestra la imágen contenida en el objeto.
        Su funcionalidad debe ser igual a la de la función mostrar_imagen.
        """
        plt.imshow(self.imagen)
        plt.show()


    def info(self):
        """ Imprime las características de la imagen cargada: Alto y ancho.
        """
        x, y, z = self.imagen.shape
        print(f'Caracteristicas de la imagen cargada: \nAlto: {x}\nAncho: {y}')


    def __add__(self, other):
        """Redefine la operación + entre imagen y escalar.

        # Idea, usar indexado condicial (similar a los filtros de pandas).
        # Sumar y luego que en cada pixel mayor a 255 sea asignado el máximo.
        # Ver los tests para mas información.

        Parameters
        ----------
        other : Union[int, float, np.ndarray]
            Escalar o arreglo que será sumado a cada pixel de la imagen
        """

        # operacion es la suma de la imagen más el escalar other
        operacion = self.imagen + other
    
        # filtro lista donde sale True o False los valores que son mayores a 255
        filtro = operacion > 255 

        # se realiza el filtro, valores mayores a 255 son iguales a 255
        operacion[filtro] = 255 

        operacion_entero = operacion.astype(int)

        return Imagen(operacion_entero)
        
    def __radd__(self, other):
        """Operación conmutativa de __add__.

        Hint: debería llamar a __add__... (porque la suma es conmutativa, es igual)

        Parameters
        ----------
        other : Union[int, float, np.ndarray]
            Escalar o arreglo que será sumado a cada pixel de la imagen
        """
        return self.__add__(other)

    def __sub__(self, other):
        """Redefine la operación - entre imagen y escalar.

        # Idea, usar indexado condicial (similar a los filtros de pandas).
        # Restar y luego que en cada pixel mayor a 255 sea asignado el máximo.
        # Caso similar para valores menores a 0, donde debera asignar el minimo a esos pixeles
        # Ver los tests para mas información.

        Parameters
        ----------
        other : Union[int, float, np.ndarray]
            Escalar o arreglo que será restado a cada pixel de la imagen
        """
        # operacion es la suma de la imagen más el escalar other
        operacion = self.imagen - other

        # filtro lista donde sale True o False los valores que son mayores a 255 y menores a 0 
        filtro1 = operacion > 255 
        filtro2 = operacion < 0

        # se realiza el filtro, valores mayores a 255 son iguales a 255
        operacion[filtro1] = 255 
        operacion[filtro2] = 0

        operacion_entero = operacion.astype(int)             

        return Imagen(operacion_entero)

    def __rsub__(self, other):
        """Operación conmutativa de __sub__.

        Parameters
        ----------
        other : Union[int, float, np.ndarray]
            Escalar o arreglo que será sumado a cada pixel de la imagen
        """
        operacion = other - self.imagen

        # filtro lista donde sale True o False los valores que son mayores a 255 y menores a 0 
        filtro1 = operacion > 255 
        filtro2 = operacion < 0

        # se realiza el filtro, valores mayores a 255 son iguales a 255
        operacion[filtro1] = 255 
        operacion[filtro2] = 0

        operacion_entero = operacion.astype(int)             

        return Imagen(operacion_entero)


    def __mul__(self, other):
        """Redefine la operación * entre imagen y escalar.

        # Idea, usar indexado condicial (similar a los filtros de pandas).
        # Sumar y luego que en cada pixel mayor a 255 sea asignado el máximo y
        # cada valor inferior a 0 debe ser 0.
        # Ver los tests para mas información.

        Parameters
        ----------
        other : Union[int, float, np.ndarray]
            Escalar o arreglo que será sumado a cada pixel de la imagen
        """
        operacion = other * self.imagen

        # filtro lista donde sale True o False los valores que son mayores a 255 y menores a 0 
        filtro1 = operacion > 255 
        filtro2 = operacion < 0

        # se realiza el filtro, valores mayores a 255 son iguales a 255
        operacion[filtro1] = 255 
        operacion[filtro2] = 0

        operacion_entero = operacion.astype(int)             

        return Imagen(operacion_entero)
    
    def __rmul__(self, other):
        """Operación conmutativa de __mul__.

        Hint: debería llamar a __mul__... (la multiplicación es conmutativa)

        Parameters
        ----------
        other : Union[int, float, np.ndarray]
            Escalar o arreglo que será sumado a cada pixel de la imagen
        """
        return self.__mul__(other)
````

### 2.3 Clase de Procesamiento de Imagenes [2 puntos]

Ahora que comprenden las diferentes dimensiones que componen a una imagen (en la práctica), ahora realizaremos diferentes tareas de procesamiento de imágenes. Para esto, deben crear una clase llamada "`LibImagen`" que cumpla los siguientes requisitos:

- [ ] Pasar una imagen a escala de grises, para esto utilice la ecuación (1) expuesta en este mismo notebook.
- [ ] Obtener los canales R, G y B de forma individual.

- [ ] Mejorar el contraste de una imagen.

> **Nota 🗒️**: Todo método debe tomar una Imagen y retornar una nueva Imagen.

> **Nota 2**: El tipo de datos del arreglo de la imagen que generen o modifiquen debe ser "int". De lo contrario, puede no visualizarse correctamente.


````python
class LibImagen():

    def to_negative(self, img_in):
        """Convierte imagen a negativo.

        Parameters
        ----------
        img_in : Imagen
            Objeto Imagen que contiene imagen a procesar.

        Returns
        -------
        Imagen
            Objeto Imagen con la imagen procesada.
        """
        img0 = img_in.imagen # asi lo definimos anteriormente en  __init__(self, img) donde self.imagen = img
        neg = 255-img0 # se resta cada valor de la imagen a 255
        neg[neg < 0] = 0 # se asigna 0 a los valores negativos
        neg[neg > 255] = 255 # se asigna 255 a los valores que superen el valor
        neg = neg.astype(int) # se convierte a tipo entero
        return Imagen(neg)


    def to_gray(self, img_in):
        """
        Transforma una imagen en RGB a la escala de grises.

        Parameters
        ----------
        img_in : Imagen
            Objeto Imagen que contiene una imagen.

        Returns
        -------
        Imagen
            Una que contiene una imagen con 3 canales.
            Los 3 canales deben tener los mismos valores.
        """
        img0 = img_in.imagen
        # Image.open('ruta') es de la libreria Pillow, por lo que el orden del array para los canales es RGB 
        r = img0[:,:,0] # canal rojo
        g = img0[:,:,1] # canal verde
        b = img0[:,:,2] # canal azul

        gray  = 0.299*r + 0.587*g +  0.114*b # se calcula la imagen en escala de grises
        gray2 = np.zeros((gray.shape[0], gray.shape[1] , 3)) # se crean las otras dimensiones con ceros

        gray2[:, :, 0] = gray # en cada canal creado se guardan los mismos valores de gris obtenido
        gray2[:, :, 1] = gray # esto permite que las dimensiones se respeten y no resente error por la definicion en  __init__ 
        gray2[:, :, 2] = gray

        gray2[gray2 < 0] = 0
        gray2[gray2 > 255] = 255
        gray2 = gray2.astype(int)

        return Imagen(gray2)
        

    def get_channel(self, img_in, channel):
        """Obtiene un canal de un color seteando el resto de los canales en 0.

        Parameters
        ----------
        img_in : Imagen
            Objeto Imagen que contiene una imagen.
        channel : str
            Nombre del canal que será seleccionado. Valores posibles: ('r','g' o 'b').

        Returns
        -------
        Imagen:
            Objeto Imagen que contiene una imagen con 3 canales.
            Solo el canal seleccionado debe tener valores distintos a 0.
        """
        img0 = img_in.imagen.copy() # esto es para no modificar el original, pues vamos a llenar de cero los canales que no indiquemos
        
        if channel == 'r':
            img0[:, :, 1] = 0  # Verde a 0
            img0[:, :, 2] = 0  # Azul a 0
        elif channel == 'g':
            img0[:, :, 0] = 0  # Rojo a 0
            img0[:, :, 2] = 0  # Azul a 0
        elif channel == 'b':
            img0[:, :, 0] = 0  # Rojo a 0
            img0[:, :, 1] = 0  # verde a 0
        
        else:
            raise ValueError("Debe indicar un canal válido: 'r', 'g' o 'b'.")

        img0[img0 < 0] = 0
        img0[img0 > 255] = 255
        img0 = img0.astype(int)
        
        return Imagen(img0)

    def set_contrast(self, img_in, C):
        """Mejora el contraste de una imagen.

        Parameters
        ----------
        img_in : Imagen
            Objeto Imagen que contiene una imagen.
        C : float
            Parámetro que define el ajuste de contraste.

        Returns
        -------
        Imagen
            Objeto Imagen que contiene una imagen con 3 canales modificados.
        """


        """
        Ajusta el contraste de la imagen con la fórmula:
        F = 259*(C+255) / (255*(259-C))
        
        """
        if not isinstance(C, (int, float)): # para verificar que se ingresa un tipo de dato válido de contraste 
            raise TypeError("C debe ser un número (int o float).")
        if C == 259: 
            raise ValueError("C debe ser distinto de 259") # evita error en división.
        
        
        img0 = img_in.imagen
        img = img0.astype(int) # por lo que aplicaremos luego una operación es preferible trabajarlo en int
        
        # Calcular factor
        F = 259 * (C + 255) / (255 * (259 - C))
        
        # Aplicar fórmula
        out = F * (img - 128) + 128
        
        
        out[out < 0] = 0
        out[out > 255] = 255
        out = out.astype(int)
        return Imagen(out)
        
    def conv_channel(self, img_in, kernel):
        """
        Aplica una convolución 2D canal por canal sobre una imagen RGB.

        Parámetros
        ----------
        img_in : Imagen
            Objeto "Imagen" que contiene la imagen original.
        kernel : np.ndarray
            Matriz 2D que define el filtro de convolución 
            (por ejemplo, para detección de bordes, suavizado, etc.).
    
        Descripción general de conv_channel(.,.,):
        -----------
        - Se separan los tres canales (R, G, B) de la imagen.
        - A cada canal se le aplica `convolve2d` con el kernel que se defina.
        - Se combinan los tres canales filtrados en una nueva imagen.
        - Se saturan los valores al rango [0,255] y se convierten a enteros.
        - Devuelve una nueva imagen con la convolución aplicada en cada canal.
        """
        img = img_in.imagen
        img_out = []                       # lista vacia donde se guardará el resultado
        for i in range(img.shape[-1]):     # img.shape obtenemos la ultima dimension (3) y luego hacemos range() que da 0,1,2
            img_channel = convolve2d(img[:, :, i], # se recorre los canales: i=0 (rojo), i=1 (verde), i=2 (azul) 
                                     kernel,       # se aplica la convolución sobre cada canal usando el kernel especificado
                                     mode="same", # hace que la imagen de salida tenga el mismo tamaño que la de entrada
                                     boundary="symm") # indica que la convolución debe reflejarse sobre sí misma al llegar a los bordes
            img_out.append(img_channel) # cada vez que se aplica convolved2D a un canal, se agrega el resultado  a la lista img_out
        new_image = np.stack(img_out, axis=2) # se combinan las tres matrices 2D a lo largo del eje 2 (se apilan los tres canales ) formando imagen 3D
        new_image[new_image>255], new_image[new_image<0] = 255, 0 # se deja en un rango definido de 0 a 255
        return Imagen(new_image.astype(int))
````
